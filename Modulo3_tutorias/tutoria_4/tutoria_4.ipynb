```python
import pandas as pd
import numpy as np
#Para el encoding
from sklearn.preprocessing import LabelEncoder #Convertir Texto ---> número entero.
```

```python
df = pd.read_csv("streaming_latam.csv")

df.head()
```

```python
#Entender y conocer tipos de datos
df.info()
```

```python
#Verificar nulos
df.isnull().sum()
```

```python
#Identificar y separar las columnas por tipo.
columnas_categoricas = df.select_dtypes(include='object').columns.tolist()

columnas_numericas = df.select_dtypes(include=['int64', 'float64']).columns.tolist()

print("Columnas Categoricas (necesitan encoding): ")
for col in columnas_categoricas:
    print(f" -> {col:25s} | {df[col].nunique()} categoricas: {df[col].unique().tolist()}")

print()
print("Columnas Numéricas (Listas para el modelo): ")
for col in columnas_numericas:
    print(f" -> {col:25s} | min={df[col].min():.1f} max={df[col].max():.1f}")
    
```

CONVERTIR 'Comparte_cuenta' -> (1/0)

```python
#Label Encoding
df_encoded = df.copy()

df_encoded['comparte_cuenta_enc'] = df_encoded['comparte_cuenta'].map({
    "Sí": 1,
    "No": 0
})

comparacion = df_encoded[['comparte_cuenta', 'comparte_cuenta_enc']].head(10)
print("Antes y después del Label Encoding Manual:")
print(comparacion.to_string(index=False))

#Verificar nulos
print(f"\n Nulos en nueva columna: {df_encoded['comparte_cuenta_enc'].isnull().sum()}")
```
