Bloque 4 - lectura SQL

```python
#Sqlite
conexion = sqlite3.connect("datajobs.db")
print("Exito")
```

```python
#Ver que tablas tenemos o tiene la base de datos

tablas = pd.read_sql(
    "SELECT name FROM sqlite_master WHERE type='table'", conexion
)

print("Tablas disponibles en datajobs.db: ")
print(tablas)
```

```python
#Traer de la db el historial de salarios

query01 = """
    SELECT * FROM historial_salarios
"""

df_salarios = pd.read_sql(query01, conexion)
df_salarios
```

```python
#Top 5 candidatos con MAYOR salario_actual
#ORDER BY
#LIMIT 8

query_top5 = """
    SELECT *
    FROM historial_salarios
    ORDER BY salario_actual DESC
    LIMIT 5
"""

df_top5 = pd.read_sql(query_top5, conexion)
print("top 5 candidatos con mayor salario: ")
df_top5
```

```python
#Candidatos que usen Python o Javascript
#IN
#WHERE

query_python_js = """
    SELECT *
    FROM historial_salarios
    WHERE tecnologia IN ('Python', 'JavaScript')
    ORDER BY salario_actual DESC
"""
df_python_js = pd.read_sql(query_python_js, conexion)
df_python_js



```
